```python
# Understanding Return Values in Functions

def add(a, b):
  return a + b

result = add(10, 20)
print(result)
```

```python
# Using Functions to Perform Calculations
def rectangle_area(width, height):
  return width * height

area = rectangle_area(10, 20)
print("The rectangle area is:", area)
```

```python
# How to Return Multiple Values
def math_operations(a, b):
  addition = a + b
  subtraction = a- b
  multiplication = a * b
  division = a / b
  return addition, subtraction, multiplication, division

add, subtract, multiply, divide = math_operations(10, 5)
print("The result of operation is: ", divide)
```

```python
# What are List Comprehensions
[expression for item in iterable if condition]
```

```python
#Basic Syntax and Examples
squares = [x**2 for x in range(10)]
print(squares)
```

```python
numbers = [1, 2, 3, 4, 5]
doubled = [x * 2 for x in numbers]
print(doubled)
```

```python
# Filtering with List Comprehensions
numbers = [1, 2, 3, 4, 5, 6, 7]
evens = [x for x in numbers if x % 2 == 0]
print(evens)
```

```python
names = ["Alice", "Bob", "Charlie", "David"]
uppercase_names = [name.upper() for name in names]
print(uppercase_names)
```

```python
names = ["Alice", "Bob", "Charlie", "Dave"]
short_names = [name for name in names if len(name) < 5]
print(short_names)
```

```python
numbers = [1, 2, 3, 4, 5, 6]
labels = ["Even" if x % 2 == 0 else "Odd" for x in numbers]
print(labels)
```
